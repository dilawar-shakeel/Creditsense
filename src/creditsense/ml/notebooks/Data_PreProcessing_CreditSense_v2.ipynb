````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

def plot_dataframe_heatmap(df, cmap='coolwarm'):
    """
    Generates a correlation heatmap with numbers on every box.

    Parameters:
    df (pd.DataFrame): The input dataframe.
    cmap (str): The color palette for the heatmap (default is 'coolwarm').
    """
    # Calculate the correlation matrix for numeric columns only
    corr_matrix = df.corr(numeric_only=True)
    num_cols = len(corr_matrix.columns)

    if num_cols == 0:
        print("No numeric columns found in the dataframe.")
        return

    # Set a larger canvas size to accommodate all the numbers comfortably
    fig_width = max(14, num_cols * 0.6)
    fig_height = max(12, num_cols * 0.5)
    plt.figure(figsize=(fig_width, fig_height))

    # Scale down the font size dynamically as the number of columns grows
    annot_font_size = max(6, 12 - (num_cols * 0.15))

    # Create the heatmap with forced annotations
    sns.heatmap(
        corr_matrix,
        cmap=cmap,
        annot=True,              # Always show numbers
        fmt=".2f",               # Format to 2 decimal places
        annot_kws={"size": annot_font_size}, # Apply the dynamic text size
        linewidths=0.3,          # Thin lines between boxes to separate values
        cbar_kws={"shrink": 0.8}
    )

    # Layout styling
    plt.title('Features Correlation Heatmap', fontsize=16, fontweight='bold', pad=20)
    plt.xticks(rotation=45, ha='right', fontsize=9)
    plt.yticks(rotation=0, fontsize=9)
    plt.tight_layout()

    # Display the plot
    plt.show()
````

## Feature Engineering


| Feature | How to build it | Why it should help |
| :--- | :--- | :--- |
| **Financial stress count** | Count how many of these cross a risky threshold:<br><br>• `late_payment_count_12m` >= 2<br>• `bank_statement_volatility` in top 25%<br>• `working_capital_cycle_days` in top 25%<br>• `utility_default_count_12m` >= 2<br><br>**Result:** A simple 0–4 score. | This is standard practice in real credit scorecards — a plain "count of red flags" is easy to explain to a risk committee or regulator, unlike a black-box combination. Combining several weak, noisy signals into one count also cancels out some of each individual signal's noise. |
| **Liquidity cushion** | `current_ratio` - (`working_capital_cycle_days` / 100)<br>*(roughly rescaled to comparable ranges)* | These two move opposite each other (-0.36 correlation) and both relate to how easily a company can cover short-term obligations — combining them into one buffer measure reflects the actual business reality better than either alone. |
| **Score-adjusted risk** | `ecib_score` - (`financial_stress_count` × 50)<br>*(or similar scaled penalty)* | `ecib_score` is your best individual predictor but is diluted by the missing-20% issue. Adjusting it down when other stress signals are present may recover some of the "cliff-effect" signal we discussed earlier — where a low score plus other red flags is a much stronger case than a low score alone. |


````python
# All thresholds calculated from TRAINING data only, then applied to both sets
bsv_75 = X_train_str['bank_statement_volatility'].quantile(0.75)
wcc_75 = X_train_str['working_capital_cycle_days'].quantile(0.75)
````

````python

def engineer_features(df):
    df = df.copy()
    df['financial_stress_count'] = (
        (df['late_payment_count_12m'] >= 2).astype(int) +
        (df['bank_statement_volatility'] >= bsv_75).astype(int) +
        (df['working_capital_cycle_days'] >= wcc_75).astype(int) +
        (df['utility_default_count_12m'] >= 2).astype(int)
    )

    df['score_adjusted_risk'] = df['ecib_score'] - (df['financial_stress_count'] * 50)
    return df
````

````python
import numpy as np
def calculate_iv(df, feature, target, bins=10):
    df = df[[feature, target]].copy()
    df['bin'] = pd.qcut(df[feature], q=bins, duplicates='drop')
    grouped = df.groupby('bin',observed=False)[target].agg(['sum', 'count'])
    grouped['non_event'] = grouped['count'] - grouped['sum']
    grouped['event_rate'] = grouped['sum'] / grouped['sum'].sum()
    grouped['non_event_rate'] = grouped['non_event'] / grouped['non_event'].sum()
    grouped['woe'] = np.log((grouped['event_rate'] + 1e-6) / (grouped['non_event_rate'] + 1e-6))
    grouped['iv'] = (grouped['event_rate'] - grouped['non_event_rate']) * grouped['woe']
    return grouped['iv'].sum()

````

````python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import scipy.stats as stats

def plot_all_numerical_distributions(data: pd.DataFrame):
    """
    Identifies all numerical columns in the DataFrame and plots 4 subplots
    (Histogram/KDE, Box Plot, Scatter Plot, Q-Q Plot) for each feature
    aligned in a single row per feature, all within one cell output.
    """
    # Select only the numerical columns
    numerical_cols = data.select_dtypes(include=['number']).columns.tolist()

    if not numerical_cols:
        print("No numerical columns found in the DataFrame.")
        return

    num_features = len(numerical_cols)

    # Setup a grid: 'num_features' rows, 4 columns
    # We scale the height dynamically (5 inches per row)
    fig, axes = plt.subplots(num_features, 4, figsize=(24, 5 * num_features))

    # Ensure axes is a 2D array even if there is only 1 numerical feature
    if num_features == 1:
        axes = axes.reshape(1, 4)

    for i, col in enumerate(numerical_cols):
        # 1. Histplot + KDE
        sns.histplot(data=data, x=col, kde=True, ax=axes[i, 0], color="skyblue")
        axes[i, 0].set_title(f'Histogram & KDE of {col}')
        axes[i, 0].set_ylabel('Frequency')

        # 2. Boxplot
        sns.boxplot(data=data, y=col, ax=axes[i, 1], color="lightgreen")
        axes[i, 1].set_title(f'Box Plot of {col}')
        axes[i, 1].set_ylabel(col)

        # 3. Scatter Plot (vs Index)
        sns.scatterplot(data=data, x=data.index, y=col, ax=axes[i, 2], alpha=0.6, color="salmon")
        axes[i, 2].set_title(f'{col} vs Index')
        axes[i, 2].set_xlabel('Index')
        axes[i, 2].set_ylabel(col)

        # 4. Q-Q Plot
        stats.probplot(data[col].dropna(), dist="norm", plot=axes[i, 3])
        axes[i, 3].set_title(f'Normal Q-Q Plot of {col}')
        # Adjusting the labels generated by scipy.stats to look cleaner
        axes[i, 3].set_xlabel('Theoretical Quantiles')
        axes[i, 3].set_ylabel('Ordered Values')

    # Prevents overlapping of text/labels across rows
    plt.tight_layout()
    plt.show()
````

### Categorical Features

````python
def plot_all_categorical_distributions(data: pd.DataFrame):
    """
    Identifies all categorical/object columns in the DataFrame and plots 2 subplots
    (Count Plot and Pie Chart) for each feature aligned in a single row per feature,
    all within one cell output. Fixed for Seaborn v0.14+ deprecation warnings.
    """
    # Select only categorical, object, or boolean columns
    categorical_cols = data.select_dtypes(include=['object', 'category', 'boolean']).columns.tolist()

    if not categorical_cols:
        print("No categorical columns found in the DataFrame.")
        return

    num_features = len(categorical_cols)

    # Setup a grid: 'num_features' rows, 2 columns (Vertical Count Plot & Pie Chart)
    fig, axes = plt.subplots(num_features, 2, figsize=(14, 5 * num_features))

    # Ensure axes is a 2D array even if there is only 1 categorical feature
    if num_features == 1:
        axes = axes.reshape(1, 2)

    for i, col in enumerate(categorical_cols):
        # Calculate frequencies and clean missing values for plotting
        value_counts = data[col].dropna().value_counts()

        if value_counts.empty:
            continue

        # 1. Countplot (Standard frequency comparison - updated to suppress warnings)
        sns.countplot(
            data=data,
            x=col,
            order=value_counts.index,
            hue=col,            # Assigned x variable to hue to fix the warning
            legend=False,       # Turned off the automatic legend
            ax=axes[i, 0],
            palette="viridis"
        )
        axes[i, 0].set_title(f'Count Plot of {col}')
        axes[i, 0].set_xlabel(col)
        axes[i, 0].set_ylabel('Count')
        axes[i, 0].tick_params(axis='x', rotation=45)

        # 2. Pie Chart (Proportional / percentage breakdown)
        axes[i, 1].pie(
            value_counts,
            labels=value_counts.index,
            autopct='%1.1f%%',
            startangle=90,
            colors=sns.color_palette("viridis", len(value_counts))
        )
        axes[i, 1].set_title(f'Proportional Share of {col}')

    # Prevents overlapping of text/labels across rows
    plt.tight_layout()
    plt.show()
````

````python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

def compare_numerical_side_by_side(df1: pd.DataFrame, name1: str, df2: pd.DataFrame, name2: str):
    """
    Compares common numerical features of two split DataFrames side-by-side.
    Generates a Scatter Plot and a Histogram/KDE plot for each feature in a single row.
    """
    # Find common numerical columns between both dataframes
    num_cols1 = df1.select_dtypes(include=['number']).columns
    num_cols2 = df2.select_dtypes(include=['number']).columns
    numerical_cols = list(set(num_cols1).intersection(num_cols2))

    if not numerical_cols:
        print("No common numerical columns found between the dataframes.")
        return

    num_features = len(numerical_cols)

    # Setup a grid: 'num_features' rows, 2 columns (Scatter and Hist/KDE)
    fig, axes = plt.subplots(num_features, 2, figsize=(14, 5 * num_features))

    # Ensure axes is a 2D array even if there is only 1 feature
    if num_features == 1:
        axes = axes.reshape(1, 2)

    for i, col in enumerate(numerical_cols):
        # Extract clean data for plotting
        d1_clean = df1[col].dropna()
        d2_clean = df2[col].dropna()

        # 1. Side-by-Side Scatter Plot (vs Index)
        sns.scatterplot(x=df1.index, y=df1[col], ax=axes[i, 0], alpha=0.5, color="black", label=name1)
        sns.scatterplot(x=df2.index, y=df2[col], ax=axes[i, 0], alpha=0.5, color="salmon", label=name2)
        axes[i, 0].set_title(f'Scatter Plot: {col} vs Index')
        axes[i, 0].set_xlabel('Index')
        axes[i, 0].set_ylabel(col)
        axes[i, 0].legend()

        # 2. Side-by-Side Bar/Histogram Plot with KDE
        sns.histplot(x=d1_clean, kde=True, ax=axes[i, 1], color="black", label=name1, alpha=0.6)
        sns.histplot(x=d2_clean, kde=True, ax=axes[i, 1], color="salmon", label=name2, alpha=0.5)
        axes[i, 1].set_title(f'Histogram & KDE: {col}')
        axes[i, 1].set_xlabel(col)
        axes[i, 1].set_ylabel('Frequency')
        axes[i, 1].legend()

    # Prevents overlapping of text/labels across rows
    plt.tight_layout()
    plt.show()
````

````python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

def compare_categorical_side_by_side(df1: pd.DataFrame, name1: str, df2: pd.DataFrame, name2: str):
    """
    Compares common categorical features of two split DataFrames side-by-side.
    Plots raw counts and normalized percentages for each feature in a single row.
    """
    # Find common categorical columns between both dataframes
    cat_cols1 = df1.select_dtypes(include=['object', 'category', 'boolean']).columns
    cat_cols2 = df2.select_dtypes(include=['object', 'category', 'boolean']).columns
    categorical_cols = list(set(cat_cols1).intersection(cat_cols2))

    if not categorical_cols:
        print("No common categorical columns found between the dataframes.")
        return

    num_features = len(categorical_cols)

    # Setup a grid: 'num_features' rows, 2 columns (Count Plot and Normalized Share Plot)
    fig, axes = plt.subplots(num_features, 2, figsize=(16, 5 * num_features))

    # Ensure axes is a 2D array even if there is only 1 feature
    if num_features == 1:
        axes = axes.reshape(1, 2)

    for i, col in enumerate(categorical_cols):
        # 1. Build a clean, combined dataframe for Seaborn grouping
        d1 = df1[[col]].copy().dropna()
        d2 = df2[[col]].copy().dropna()
        d1['Class'] = name1
        d2['Class'] = name2
        combined = pd.concat([d1, d2])

        if combined.empty:
            continue

        # Determine universal order based on highest total frequency
        order = combined[col].value_counts().index

        # 2. Calculate percentages to avoid bias if dataframes have different lengths
        prop_df = (
            combined.groupby('Class')[col]
            .value_counts(normalize=True)
            .rename('Percentage')
            .reset_index()
        )
        prop_df['Percentage'] *= 100  # Convert fractions to 0-100%

        # Plot 1: Side-by-Side Count Plot (Raw Volumes)
        sns.countplot(
            data=combined,
            x=col,
            hue='Class',
            order=order,
            ax=axes[i, 0],
            palette=["skyblue", "salmon"]
        )
        axes[i, 0].set_title(f'Raw Counts: {col}')
        axes[i, 0].set_xlabel(col)
        axes[i, 0].set_ylabel('Count')
        axes[i, 0].tick_params(axis='x', rotation=45)
        axes[i, 0].legend(title='DataFrame')

        # Plot 2: Side-by-Side Normalized Bar Plot (Relative Share Percentage)
        sns.barplot(
            data=prop_df,
            x=col,
            y='Percentage',
            hue='Class',
            order=order,
            ax=axes[i, 1],
            palette=["skyblue", "salmon"]
        )
        axes[i, 1].set_title(f'Proportional Share (%): {col}')
        axes[i, 1].set_xlabel(col)
        axes[i, 1].set_ylabel('Percentage (%)')
        axes[i, 1].tick_params(axis='x', rotation=45)
        axes[i, 1].legend(title='DataFrame')

    # Prevents overlapping of text/labels across rows
    plt.tight_layout()
    plt.show()
````

````python
from __future__ import annotations

import json
import warnings
from dataclasses import dataclass, field

import joblib
import numpy as np
import pandas as pd
import shap
from sklearn.calibration import CalibratedClassifierCV
from sklearn.frozen import FrozenEstimator
from sklearn.metrics import (
    average_precision_score,
    classification_report,
    mean_absolute_error,
    r2_score,
)
from sklearn.model_selection import (
    RandomizedSearchCV,
    StratifiedKFold,
    cross_val_predict,
    train_test_split,
)
from sklearn.preprocessing import OneHotEncoder
from xgboost import XGBClassifier, XGBRegressor

RANDOM_STATE = 42
warnings.filterwarnings("ignore", category=FutureWarning)
````

# 1. FEATURE LISTS

````python
RAW_FEATURES_STAGE1 = [
    "current_ratio",
    "debt_to_equity_ratio",
    "revenue_growth_yoy_pct",
    "sector_risk_code",
    "years_in_business",
    "existing_loan_exposure_pkr",
    "late_payment_count_12m",
    "collateral_coverage_ratio",
    "bank_statement_volatility",
    "ecib_score",
    "kibor_sensitivity_pct",
    "documentation_tier",
    "working_capital_cycle_days",
    "utility_default_count_12m",
]
TARGET_1 = "default_probability_12m"

STAGE2_EXTRA_FEATURES = [
    "annual_bank_turnover_pkr",
    "guarantor_net_worth_pkr",
    "group_associate_exposure_pkr",
    "sbp_enterprise_tier",
]
TARGET_2 = "recommend_credit_limit_pkr"

# Direction each feature is ALLOWED to push risk, left-to-right in the sense of
# "as this number goes up, predicted default probability may only ..."
#   +1 = ... go up or stay flat (never down)
#   -1 = ... go down or stay flat (never up)
#   feature absent from this dict = no constraint (direction is genuinely unclear
#   or the feature is categorical, where "up/down" has no meaning)
MONOTONE_DIRECTIONS = {
    "current_ratio": -1,             # more short-term liquidity -> risk may not rise
    "debt_to_equity_ratio": 1,        # more leverage -> risk may not fall
    "collateral_coverage_ratio": -1,  # more collateral behind the loan -> risk may not rise
    "late_payment_count_12m": 1,      # more late payments -> risk may not fall
    "utility_default_count_12m": 1,   # more utility defaults -> risk may not fall
    "existing_loan_exposure_pkr": 1,  # more existing debt -> risk may not fall
    "financial_stress_count": 1,      # more red flags -> risk may not fall
    "score_adjusted_risk": -1,        # higher (safer) blended score -> risk may not rise
}


def build_monotone_constraints(columns: list[str]) -> tuple[int, ...]:
    """Turn the business rules above into the exact tuple XGBoost expects,
    lined up with whatever column order the data happens to be in."""
    return tuple(MONOTONE_DIRECTIONS.get(c, 0) for c in columns)
````

# 2. STAGE 1 PREPROCESSING


````python

class Stage1Preprocessor:
    def __init__(self):
        self.bsv_75_: float | None = None
        self.wcc_75_: float | None = None
        self.onehot_ = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
        self._fitted = False

    def fit(self, df: pd.DataFrame) -> "Stage1Preprocessor":
        self.bsv_75_ = df["bank_statement_volatility"].quantile(0.75)
        self.wcc_75_ = df["working_capital_cycle_days"].quantile(0.75)
        self.onehot_.fit(df[["documentation_tier"]])
        self._fitted = True
        return self

    def transform(self, df: pd.DataFrame) -> pd.DataFrame:
        if not self._fitted:
            raise RuntimeError("Call .fit() on the TRAINING set before .transform().")

        out = df[RAW_FEATURES_STAGE1].copy()

        # -- Engineered feature 1: how many independent stress signals fire at once.
        # Thresholds are fixed numbers learned from training data, not recalculated here.
        out["financial_stress_count"] = (
            (out["late_payment_count_12m"] >= 2).astype(int)
            + (out["bank_statement_volatility"] >= self.bsv_75_).astype(int)
            + (out["working_capital_cycle_days"] >= self.wcc_75_).astype(int)
            + (out["utility_default_count_12m"] >= 2).astype(int)
        )

        # -- Engineered feature 2: bureau score, discounted by how many stress
        # signals are firing. Tested to clearly outperform either ingredient alone.
        out["score_adjusted_risk"] = out["ecib_score"] - (out["financial_stress_count"] * 50)

        # Plain ecib_score dropped: tested weak on its own (Information Value ~0.02,
        # even when missing values are treated as their own group) and its useful
        # signal is already carried inside score_adjusted_risk.
        out = out.drop(columns=["ecib_score"])

        # documentation_tier: three tiers with NO natural ranking (bank-statement-only
        # applicants tested riskier than unaudited-financials applicants, so a
        # 0/1/2 ranking would have been actively wrong) -> one-hot encode.
        enc = self.onehot_.transform(out[["documentation_tier"]])
        enc_cols = self.onehot_.get_feature_names_out(["documentation_tier"])
        enc_df = pd.DataFrame(enc, columns=enc_cols, index=out.index)
        out = pd.concat([out.drop(columns=["documentation_tier"]), enc_df], axis=1)

        # sector_risk_code: also no natural ranking -> native pandas category,
        # handled directly by XGBoost (avoids double the columns one-hot would add).
        out["sector_risk_code"] = out["sector_risk_code"].astype("category")

        return out

    def fit_transform(self, df: pd.DataFrame) -> pd.DataFrame:
        return self.fit(df).transform(df)

````

# 3. STAGE 1 MODEL

````python
def tune_stage1(X_train: pd.DataFrame, y_train: pd.Series, monotone_constraints,
                 n_iter: int = 25, cv_folds: int = 5) -> tuple[dict, float]:
    """Searches for good XGBoost settings using cross-validated PR-AUC
    (never plain accuracy — with ~90% of companies non-defaulters, accuracy
    rewards a model that just guesses 'safe' every time)."""

    scale_pos_weight = (y_train == 0).sum() / (y_train == 1).sum()

    base_model = XGBClassifier(
        scale_pos_weight=scale_pos_weight,
        eval_metric="aucpr",
        enable_categorical=True,
        tree_method="hist",
        monotone_constraints=monotone_constraints,
        random_state=RANDOM_STATE,
    )

    param_grid = {
        "max_depth": [3, 4, 5],
        "learning_rate": [0.01, 0.03, 0.05, 0.1],
        "n_estimators": [200, 300, 400, 600],
        "min_child_weight": [1, 3, 5, 10],
        "subsample": [0.7, 0.8, 0.9, 1.0],
        "colsample_bytree": [0.7, 0.8, 0.9, 1.0],
    }

    cv = StratifiedKFold(n_splits=cv_folds, shuffle=True, random_state=RANDOM_STATE)
    search = RandomizedSearchCV(
        base_model,
        param_grid,
        n_iter=n_iter,
        scoring="average_precision",  # PR-AUC
        cv=cv,
        random_state=RANDOM_STATE,
        n_jobs=-1,
        verbose=1,
    )
    search.fit(X_train, y_train)  # training data only — no other variable is ever touched here
    return search.best_params_, scale_pos_weight


def fit_stage1_calibrated(X_train: pd.DataFrame, y_train: pd.Series, best_params: dict,
                           scale_pos_weight: float, monotone_constraints):
    """Fits the real model on most of the training data, then fits Platt scaling
    (sigmoid calibration) on a slice the model has never been trained on. Both
    slices come from the TRAINING set only — the test set stays untouched until
    final evaluation."""

    X_core, X_calib, y_core, y_calib = train_test_split(
        X_train, y_train, test_size=0.2, stratify=y_train, random_state=RANDOM_STATE
    )

    core_model = XGBClassifier(
        **best_params,
        scale_pos_weight=scale_pos_weight,
        eval_metric="aucpr",
        enable_categorical=True,
        tree_method="hist",
        monotone_constraints=monotone_constraints,
        random_state=RANDOM_STATE,
    )
    core_model.fit(X_core, y_core)

    calibrated_model = CalibratedClassifierCV(FrozenEstimator(core_model), method="sigmoid")
    calibrated_model.fit(X_calib, y_calib)
    return calibrated_model


def get_raw_xgb_model(calibrated_model: CalibratedClassifierCV) -> XGBClassifier:
    """Reaches past the calibration wrapper to the underlying tree model —
    needed for SHAP and for feature-importance checks."""
    return calibrated_model.calibrated_classifiers_[0].estimator.estimator


def recommend_threshold(y_true, y_proba) -> tuple[float, float]:
    """Finds the cutoff with maximum separation between defaulters and
    non-defaulters (the KS statistic) — the standard way credit-scoring teams
    pick a cutoff, rather than defaulting to a plain 0.5."""
    order = np.argsort(y_proba)
    scores = np.asarray(y_proba)[order]
    actual = np.asarray(y_true)[order]
    cum_bad = np.cumsum(actual == 1) / max((actual == 1).sum(), 1)
    cum_good = np.cumsum(actual == 0) / max((actual == 0).sum(), 1)
    ks_gap = np.abs(cum_bad - cum_good)
    best_idx = np.argmax(ks_gap)
    return float(scores[best_idx]), float(ks_gap[best_idx])
````

# 4. SHAP EXPLANATIONS — plain-English reason codes


````python
def explain_decision(calibrated_model: CalibratedClassifierCV, X_row: pd.DataFrame,
                      top_n: int = 3) -> str:
    """Returns a short, plain-English explanation of the top drivers behind
    one applicant's score. X_row must be a single-row DataFrame with the same
    columns the model was trained on."""

    raw_model = get_raw_xgb_model(calibrated_model)
    explainer = shap.TreeExplainer(raw_model)
    shap_values = explainer.shap_values(X_row)
    if isinstance(shap_values, list):  # older SHAP versions return a list for binary tasks
        shap_values = shap_values[1]

    contributions = pd.Series(np.ravel(shap_values), index=X_row.columns)
    contributions = contributions.reindex(contributions.abs().sort_values(ascending=False).index)

    lines = []
    for feature, value in contributions.head(top_n).items():
        direction = "pushed the risk score up" if value > 0 else "pulled the risk score down"
        raw_value = X_row[feature].values[0]
        if isinstance(raw_value, (np.floating, np.integer)):
            raw_value = raw_value.item()
        lines.append(f"  - {feature} = {raw_value}  ({direction})")

    return "Main reasons behind this score:\n" + "\n".join(lines)
````

# 5. STAGE 2 — recommended credit limit


````python
def build_stage2_features(stage1_processed: pd.DataFrame, raw_df: pd.DataFrame,
                           stage1_pd: np.ndarray, onehot_tier_encoder: OneHotEncoder | None = None,
                           fit_encoder: bool = False) -> tuple[pd.DataFrame, OneHotEncoder]:
    """Combines Stage 1's processed features with Stage 1's own predicted
    probability and the extra Stage 2-only inputs (turnover, guarantor net
    worth, group exposure, enterprise tier)."""

    out = stage1_processed.copy()
    out["stage1_predicted_pd"] = stage1_pd

    for col in ["annual_bank_turnover_pkr", "guarantor_net_worth_pkr", "group_associate_exposure_pkr"]:
        out[col] = raw_df[col].values

    if onehot_tier_encoder is None:
        onehot_tier_encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
    if fit_encoder:
        onehot_tier_encoder.fit(raw_df[["sbp_enterprise_tier"]])

    enc = onehot_tier_encoder.transform(raw_df[["sbp_enterprise_tier"]])
    enc_cols = onehot_tier_encoder.get_feature_names_out(["sbp_enterprise_tier"])
    enc_df = pd.DataFrame(enc, columns=enc_cols, index=out.index)
    out = pd.concat([out, enc_df], axis=1)

    return out, onehot_tier_encoder


def train_stage2_regressor(X_train_s2: pd.DataFrame, y_train_s2: pd.Series) -> XGBRegressor:
    monotone = build_monotone_constraints(list(X_train_s2.columns))
    # extend with the two Stage-2-specific business rules
    monotone_dict = dict(zip(X_train_s2.columns, monotone))
    monotone_dict["annual_bank_turnover_pkr"] = 1     # more verified turnover -> limit may not fall
    monotone_dict["stage1_predicted_pd"] = -1         # higher default risk -> limit may not rise
    monotone = tuple(monotone_dict[c] for c in X_train_s2.columns)

    model = XGBRegressor(
        n_estimators=400,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        enable_categorical=True,
        tree_method="hist",
        monotone_constraints=monotone,
        random_state=RANDOM_STATE,
    )
    model.fit(X_train_s2, y_train_s2)
    return model
````

# 6. SAVE / LOAD  — a real deployment needs this, a notebook usually skips it


````python

@dataclass
class CreditSenseBundle:
    stage1_preprocessor: Stage1Preprocessor
    stage1_model: CalibratedClassifierCV
    stage1_threshold: float
    stage2_tier_encoder: OneHotEncoder
    stage2_model: XGBRegressor
    metadata: dict = field(default_factory=dict)

    def save(self, path: str):
        joblib.dump(self, path)

    @staticmethod
    def load(path: str) -> "CreditSenseBundle":
        return joblib.load(path)

    def score_applicant(self, raw_row: pd.DataFrame) -> dict:
        """End-to-end scoring for ONE new applicant. `raw_row` must be a
        single-row DataFrame with all the raw columns the model expects."""
        x1 = self.stage1_preprocessor.transform(raw_row)
        pd_estimate = self.stage1_model.predict_proba(x1)[:, 1]
        decision = "DECLINE" if pd_estimate[0] >= self.stage1_threshold else "REFER_FOR_LIMIT"

        result = {
            "default_probability": float(pd_estimate[0]),
            "decision_cutoff": self.stage1_threshold,
            "decision": decision,
            "explanation": explain_decision(self.stage1_model, x1),
        }

        if decision == "REFER_FOR_LIMIT":
            x2, _ = build_stage2_features(
                x1, raw_row, pd_estimate, onehot_tier_encoder=self.stage2_tier_encoder, fit_encoder=False
            )
            x2 = x2.reindex(columns=self.metadata["stage2_columns"], fill_value=0)
            result["recommended_credit_limit_pkr"] = float(self.stage2_model.predict(x2)[0])

        return result
````

# 7. END-TO-END TRAINING RUN


````python

def main(csv_path: str = "creditsense_synthetic_portfolio.csv",
         tuning_iterations: int = 25, cv_folds: int = 5,
         output_path: str = "creditsense_bundle.joblib") -> CreditSenseBundle:

    df = pd.read_csv(csv_path)

    # ---- STAGE 1 : split, stratified on the TARGET (not on any single feature) ----
    train_df, test_df = train_test_split(
        df, test_size=0.2, random_state=RANDOM_STATE, stratify=df[TARGET_1]
    )

    pre = Stage1Preprocessor()
    X_train_1 = pre.fit_transform(train_df)
    X_test_1 = pre.transform(test_df)
    y_train_1 = train_df[TARGET_1]
    y_test_1 = test_df[TARGET_1]

    monotone = build_monotone_constraints(list(X_train_1.columns))

    print("Tuning Stage 1 (this can take a few minutes)...")
    best_params, scale_pos_weight = tune_stage1(
        X_train_1, y_train_1, monotone, n_iter=tuning_iterations, cv_folds=cv_folds
    )
    print("Best Stage 1 settings:", best_params)

    stage1_model = fit_stage1_calibrated(
        X_train_1, y_train_1, best_params, scale_pos_weight, monotone
    )

    test_proba = stage1_model.predict_proba(X_test_1)[:, 1]
    threshold, ks_score = recommend_threshold(y_test_1, test_proba)
    print(f"\nStage 1 test PR-AUC: {average_precision_score(y_test_1, test_proba):.4f}")
    print(f"Stage 1 KS score: {ks_score:.4f} | recommended cutoff: {threshold:.4f}")
    print(classification_report(y_test_1, (test_proba >= threshold).astype(int)))

    # ---- STAGE 2 : own predicted PD as an input, generated out-of-fold on TRAIN ----
    print("\nGenerating out-of-fold Stage 1 probabilities for Stage 2 training...")
    oof_model = XGBClassifier(
        **best_params,
        scale_pos_weight=scale_pos_weight,
        eval_metric="aucpr",
        enable_categorical=True,
        tree_method="hist",
        monotone_constraints=monotone,
        random_state=RANDOM_STATE,
    )
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
    train_oof_pd = cross_val_predict(oof_model, X_train_1, y_train_1, cv=cv, method="predict_proba")[:, 1]
    test_pd_for_stage2 = stage1_model.predict_proba(X_test_1)[:, 1]  # the real, final model for test/production

    X_train_2, tier_encoder = build_stage2_features(
        X_train_1, train_df, train_oof_pd, fit_encoder=True
    )
    X_test_2, _ = build_stage2_features(
        X_test_1, test_df, test_pd_for_stage2, onehot_tier_encoder=tier_encoder, fit_encoder=False
    )
    y_train_2 = train_df[TARGET_2]
    y_test_2 = test_df[TARGET_2]

    stage2_model = train_stage2_regressor(X_train_2, y_train_2)
    test_pred_2 = stage2_model.predict(X_test_2)
    print(f"\nStage 2 test R^2: {r2_score(y_test_2, test_pred_2):.4f}")
    print(f"Stage 2 test MAE: {mean_absolute_error(y_test_2, test_pred_2):,.0f} PKR")

    bundle = CreditSenseBundle(
        stage1_preprocessor=pre,
        stage1_model=stage1_model,
        stage1_threshold=threshold,
        stage2_tier_encoder=tier_encoder,
        stage2_model=stage2_model,
        metadata={"stage2_columns": list(X_train_2.columns), "ks_score": ks_score},
    )
    bundle.save(output_path)
    print(f"\nSaved trained bundle to {output_path}")

    # ---- quick demo of a single applicant explanation ----
    sample_row = test_df.iloc[[0]]
    print("\n--- Sample applicant decision ---")
    print(json.dumps({k: v for k, v in bundle.score_applicant(sample_row).items() if k != "explanation"}, indent=2))
    print(bundle.score_applicant(sample_row)["explanation"])

    return bundle


if __name__ == "__main__":
    main()
````
